```python
plt.figure(figsize=(8,4))
plt.hist(df1['price'], bins=50)
plt.xlim(0,10000)
plt.axvline(df1['price'].mean(), color='red',
            linestyle='--', linewidth=1.5,
            label=f"Mean:   ${df['price'].mean():.0f}")

plt.axvline(df1['price'].median(), color='orange',
            linestyle='--', linewidth=1.5,
            label=f"Median: ${df1['price'].median():.0f}")

plt.xlabel('Price per night ($)')
plt.ylabel('Number of listings')
plt.title('Price distribution — Airbnb NYC')
plt.legend()
plt.tight_layout()
plt.show()
```

```python
sns.boxplot(x=df1['price'])
plt.show()
```

```python
Q3 = df1['price'].quantile(0.75)
Q1 = df1['price'].quantile(0.25)
IQR = Q3 - Q1
lower = 0
upper = Q3 + 1.5 * IQR
print(lower, upper)
```

```python
clean_df = df1[(df1['price'] > 0) & (df1['price'] < upper)]
```

```python
plt.hist(df1['price'],bins=50)
plt.xlabel("Price")
plt.ylabel("Number of listing")
clean_df['price'].skew()
```

```python
plt.hist(clean_df['price'],bins=50)
```

```python
clean_df.shape
```

```python
clean_df['last_review'] = (
    clean_df['last_review']
    .replace('unknown',pd.NA)
    .pipe(pd.to_datetime,errors = 'coerce')
)
```

```python
plt.figure(figsize=(6,4))
ax = sns.barplot(clean_df['neighbourhood_group'].value_counts(),color='teal')

for container in ax.containers:
    ax.bar_label(container)
plt.ylabel('Number of counts')
```

```python
plt.figure(figsize=(8,5))
sns.boxplot(x='room_type' , y='price', data = clean_df)
plt.title("Price distribution by room type")
plt.show()
```

```python
ax3 = sns.barplot(x="room_type",y="price", data = clean_df)
for container in ax3.containers:
    ax3.bar_label(container)
plt.title("Average Price by room type")
plt.show()
```

```python
ax2 = sns.countplot(x='room_type', data=clean_df)
for container in ax2.containers:
    ax2.bar_label(container)
    
plt.title("Number of Listing by Room types")
plt.show()
```

```python
clean_df= clean_df.rename(columns={'neighbourhood_group' : 'nei_group',
                                   'availability_365': 'availablity'
                                  })
stats =clean_df.groupby('nei_group')['price'].describe().T
stats = stats.loc[['min', '25%' , '50%' , '75%', 'max']]
stats
```

#Listing Map with Prices

```python
import plotly.express as px
import plotly.io as pio
pio.renderers.default = 'notebook'

fig = px.scatter_map(
    clean_df,
    lon='longitude',
    lat='latitude',
    color=np.log1p(clean_df['price']),
    color_continuous_scale='turbo',
    title='Listing map',
    zoom=10,
    hover_data= {
        'longitude': False,
        'latitude' : False,
        'price' : True,
        'room_type':True,
        'availablity': True,
        'nei_group': True
    },
)

fig.update_layout(map_style="carto-darkmatter",width=1000)
fig.show()
```

```python
bar = sns.barplot(data = clean_df, x='nei_group', y ='price',hue='room_type')
bar.set(xlabel='Room Type', ylabel='Price', title='Room Type Vs Price in different Neighbourhood Groups')
```

```python
counts = clean_df.nei_group.value_counts()
plt.figure(figsize=(6,6))
plt.pie(
    counts,
    labels=counts.index,
    autopct='%1.1f%%'
)

plt.title("Neighbourhood Group Distribution")
plt.legend()
plt.show()
```
